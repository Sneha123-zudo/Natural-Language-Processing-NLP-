HTML TAGS Cleaning

```python
sample_text = '<p>The HTML <code>button</code> tag defines a clickable button.</p><p>The CSS <code>background-color</code> property defines the background color of an element.</p>'
```

```python
sample_text
```

```python
import re
def cleanhtml(data):
  p = re.compile(r'<.*?>')
  return p.sub('',data)
```

```python
cleanhtml(sample_text)
```

Tokenization

```python
text = 'Life is a long story made of many small moments. Some days are easy and some days are hard, but every day teaches you something new. You may fail, you may feel tired, and you may want to stop, but each step you take still matters. Even slow progress is still progress.'
```

```python
text
```

```python
import nltk
```

```python
from nltk.tokenize import sent_tokenize, word_tokenize
sents = sent_tokenize(text)
```

```python
sents
```

```python
for sent in sents:
  print(word_tokenize(sent))
```

```python
words = word_tokenize(text)
words
```
