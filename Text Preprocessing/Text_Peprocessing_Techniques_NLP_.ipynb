```python
text = "<p>I am Learning NLP!!! Visit www.site.com 😊 I cant w8 to lern more.</p>"
```

**Lowercasing - It treast every text as same**

```python
text = text.lower()
text
```

**Remove Tags - HTML**

```python
import re
text = re.sub(r'<.*?>', '', text)
print(text)
```

**Remove URLs**

```python
text = re.sub(r'http\S+|www\S+', '', text)
print(text)
```

**Expand Contractions**

```python
text = text.replace("cant", "cannot")
text
```

**Chat word treatment**

```python
chat_dict = {"w8":"wait","lern":"learn"}
words = text.split()
text = " ".join([chat_dict.get(w,w) for w in words])
print(text)
```

**Removing Emojis**

```python
text = re.sub(r'[^\w\s]', '', text)
print(text)
```

**Remove numbers**

```python
text = re.sub(r'\d+', '', text)
print(text)
```

**Remove punctutaions**

```python
text = re.sub(r'[^\w\s]', '', text)
print(text)
```

**Remove extra spaces**

```python
text = re.sub(r'\s+', ' ', text).strip()
print(text)
```

**Tokenization**

```python
import nltk
from nltk.tokenize import word_tokenize
nltk.download('punkt')
nltk.download('punkt_tab') # Added to download the missing resource

tokens = word_tokenize(text)
print(tokens)
```

**Remove Stop words**

```python
from nltk.corpus import stopwords
nltk.download('stopwords')
stop_words = set(stopwords.words('english'))
tokens = [w for w in tokens if not w in stop_words]
print(tokens)
```

**Stemming - reduce word form**

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
stemmed = [stemmer.stem(w) for w in tokens]
print(stemmed)
```

**Lemmatization - reduce in base forms**

```python
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
Lemmatizer = WordNetLemmatizer()
Lemmatized = [Lemmatizer.lemmatize(w) for w in tokens]
print(Lemmatized)
```
