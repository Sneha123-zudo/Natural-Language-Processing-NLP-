**NLP Library - Spacy**

```python
import spacy
```

```python
nlp = spacy.load('en_core_web_sm')
doc = nlp("Apple is looking at buying U.K. startup for $1 billion. AI/ML is the Trending topic nowdays")
```

Sentence Tokenization

```python
for sentence in doc.sents:
  print(sentence)
```

Word Tokenization

```python
for sentence in doc.sents:
  for word in sentence:
    print(word)
```

**NLP Library - NLTK**

```python
import nltk
nltk.download('punkt_tab')
```

```python
from nltk import sent_tokenize
sent_tokenize("Apple is looking at buying U.K. startup for $1 billion. AI/ML is the Trending topic nowdays")
```

```python
from nltk import word_tokenize
word_tokenize("Apple is looking at buying U.K. startup for $1 billion. AI/ML is the Trending topic nowdays")
```
